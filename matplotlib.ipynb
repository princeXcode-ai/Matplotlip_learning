```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Matplotlib Basics

```python
x  = np.linspace(0,6,12)
x
```

```python
y = x**2
y
```

```python
plt.plot(x,y)
plt.title("Graph of square function")
plt.xlabel("X axis")
plt.ylabel("Y axis")
```

```python
plt.subplot(2,2,2)
plt.plot(y,x)
plt.subplot(2,2,1)
plt.plot(x,x)
plt.subplot(2,2,4)
plt.plot(x,y**2)
plt.subplot(2,2,3)
plt.plot(x,y)
```

Object oriented way

```python
fig = plt.figure(figsize=(6,7),dpi =100)
axis1= fig.add_axes([0.1,0.1,1,1])
axis1.plot(x,y)
axis2 = fig.add_axes([0.3,0.4,.4,.3])
axis2.plot(y,x)
```
